2 - Given a list of words: words = ["apple", "banana", "kiwi", "cherry", "mango"]
Create a dictionary that maps each word to its corresponding length.
Example Output: ({"apple": 5, "banana": 6, "kiwi": 4, "cherry": 6, "mango": 5})


```python
# List of Words

words = ["apple", "banana", "kiwi", "cherry", "mango"]

word_lengths = {}

for word in words:
    word_lengths[word] = len(word)

print(word_lengths)
```

3 - Create a Chat System using Object-Oriented Programming (OOP) concepts.
You need to create the following classes:
*   User
*   Message
*   ChatRoom

The system should implement the following functionalities:

*   Sending messages
*   Viewing chat history
*   User joining and leaving the chat room



```python
# User Class
class User:
    def __init__(self, name):
        self.name = name

    def send_message(self, chatroom, content):
        message = Message(self, content)
        chatroom.add_message(message)

    def join_chatroom(self, chatroom):
        chatroom.add_user(self)

    def leave_chatroom(self, chatroom):
        chatroom.remove_user(self)

# Message Class

class Message:
    def __init__(self, sender, content):
        self.sender = sender
        self.content = content

    def display(self):
        return f"{self.sender.name}: {self.content}"


# ChatRoom Class
class ChatRoom:
    def __init__(self):
        self.users = []
        self.messages = []

    def add_user(self, user):
        self.users.append(user)
        print(f"{user.name} joined the chat.")

    def remove_user(self, user):
        self.users.remove(user)
        print(f"{user.name} left the chat.")

    def add_message(self, message):
        if message.sender in self.users:
            self.messages.append(message)
        else:
            print(f"{message.sender.name} is not in the chat room!")

    def show_chat_history(self):
        print("\nChat History:")
        for msg in self.messages:
            print(msg.display())

# Example Usage
chat = ChatRoom()

user1 = User("Suhail")
user2 = User("Aslam")

user1.join_chatroom(chat)
user2.join_chatroom(chat)

user1.send_message(chat, "Asalam O Alikum Aslam!")
user2.send_message(chat, "Walikum Salam Suhail, how are you?")

user1.send_message(chat, "I am fine ")

user1.send_message(chat, "Did you Complete Your Excercise 1 ?")
user2.send_message(chat, "Yes Completed and Just Submitting")
user2.send_message(chat, "Ok will chat after submission.... Allah Hafiz")

chat.show_chat_history()

user2.leave_chatroom(chat)
```
